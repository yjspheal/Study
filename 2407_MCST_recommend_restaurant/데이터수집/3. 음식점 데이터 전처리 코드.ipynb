```python
# 카테고리 매핑 딕셔너리 생성
category_mapping = {
    '중식': ['중식당', '딤섬,중식만두', '마라탕'],
    '커피/차': ['카페,디저트', '북카페', '테이크아웃커피', '카페'],
    '한식': ['문래돼지불백', '돼지고기구이', '국밥', '백반,가정식', '덮밥', '칼국수,만두', 
        '향토음식', '찌개,전골', '순대,순댓국', '한식', '해장국', '닭발', '45년의정부부대찌개', 
        '전,빈대떡', '기사식당', '한식뷔페'],
    '버거': ['햄버거'],
    '디저트': ['와플', '케이크전문', '아이스크림', '베이커리', '과일,주스전문점', '빙수'],
    '치킨': ['치킨,닭강정', '닭갈비', '닭요리'],
    '분식': ['김밥', '떡볶이', '종합분식', '분식', '국수', '라면'],
    '돈까스': ['돈가스'],
    '족발/보쌈': ['족발,보쌈'],
    '찜/탕': ['아귀찜,해물찜', '매운탕,해물탕', '찜닭', '백숙,삼계탕', '곰탕,설렁탕', '추어탕', '감자탕', '닭볶음탕'],
    '구이': ['소고기구이', '양갈비', '양꼬치', '곱창,막창,양', '정육식당'],
    '피자': ['피자'],
    '일식': ['이자카야', '일식튀김,꼬치', '초밥,롤', '일본식라면', '일식당',
           '우동,소바', '오뎅,꼬치', '오므라이스', '샤브샤브']
    '회/해물': ['대게요리', '생선회', '해물,생선요리', '주꾸미요리', '조개요리', '복어요리', '낙지요리',
            '장어,먹장어요리', '게요리'],
    '양식': ['스테이크,립', '프랑스음식', '이탈리아음식', '브런치', '스파게티,파스타전문', 
           '양식', '패밀리레스토랑', '스페인음식'],
    '간식': ['토스트', '후렌치후라이', '호떡', '만두'],
    '아시안': ['베트남음식', '태국음식', '인도음식', '아시아음식'],
    '도시락': ['도시락,컵밥'],
    '샌드위치': ['샌드위치'],
    '샐러드': ['다이어트,샐러드'],
    '멕시칸': ['멕시코,남미음식'],
    '죽': ['죽'],
    '주류': ['맥주,호프', '와인', '바(BAR)', '요리주점', '전통,민속주점', '술집', '포장마차'],
    '퓨전음식': ['퓨전음식']
}

service_items = ['포장', '핸드드립', '출입구 휠체어 이용가능', '남/녀 화장실 구분', 
                 '장애인 휠체어 이용가능', '로스터리', '방문접수/출장', '무선 인터넷', 
                 '좌석 휠체어 이용가능', '단체 이용 가능', '콜키지 가능', '발렛파킹', 
                 '유아시설 (놀이방)', '글루텐프리 메뉴', '대기공간', '예약', '유아의자', 
                 '노키즈존', '테이크아웃 할인', '비건 메뉴', '생일 혜택', '와인 페어링', 
                 '무한 리필', '주차', '유기농 메뉴', '반려동물 동반', '간편결제', '드라이브스루', 
                 '장애인 편의시설', '배달', '장애인 주차구역', '전문 소믈리에']

final_df['services'] = final_df['services'].replace('Not found', '')
final_df['seats'] = final_df['seats'].replace('Not found', '')

parking_items = ['주차가능유료', '주차불가', '주차가능', '주차가능무료']

# 각 카테고리를 매핑하여 새로운 분류 기준을 생성
def map_category(category):
    for new_cat, old_cats in category_mapping.items():
        if category in old_cats:
            return new_cat
    return '기타'  # 매핑되지 않는 경우 '기타'로 분류

# 숫자를 제거하고 '미터'를 제거하는 함수
def clean_distance(text):
    if pd.isna(text):
        return text
    text = str(text)  # 문자열로 변환
    # 가장 앞의 숫자 제거
    text = re.sub(r'^\d+', '', text).strip()
    # 끝의 '미터' 제거
    text = re.sub(r'미터$', '', text).strip()
    return text

# href 속성에서 쌍따옴표 안의 링크를 추출하는 함수
def extract_href_links(html):
    if pd.isna(html):
        return html
    href_pattern = r'href="([^"]*)"'
    href_texts = re.findall(href_pattern, html)
    return href_texts

def merge_and_remove_duplicates(row):
    # services와 home_service 칼럼을 합치고 ','로 분리하여 중복 제거 후 다시 합침
    services = row['services'] if pd.notna(row['services']) else ''
    home_service = row['home_service'] if pd.notna(row['home_service']) else ''
    
    # 빈 문자열을 NaN으로 처리하고, NaN은 빈 문자열로 변환
    combined = (services + ', ' + home_service).replace(', nan', '')
    unique_services = list(set(combined.split(', ')))
    return ', '.join(unique_services)

# 부가 서비스 원핫인코딩
def create_service_columns(df, service_items):
    # 각 서비스 항목에 대한 칼럼 만들기
    for item in service_items:
        df[item] = df['unique_services'].apply(lambda x: 1 if isinstance(x, str) and item in x.split(', ') else 0)
    
    return df

# 주차 서비스 원핫인코딩
def create_parking_columns(df, parking_items):
    # 각 서비스 항목에 대한 칼럼 만들기
    for item in parking_items:
        df[item] = df['parking'].apply(lambda x: 1 if isinstance(x, str) and item in x.split(', ') else 0)
    
    return df

# 메뉴 데이터 분리(메뉴 정보/메뉴판 이미지)
def extract_menu_info_and_img(data_str):
    try:
        # 문자열을 파이썬 객체로 변환 (튜플로 변환)
        data_tuple = ast.literal_eval(data_str)
        # 첫 번째 리스트 추출 (menu_info)
        menu_info = data_tuple[0]
        # 두 번째 리스트 추출 (menu_img)
        menu_img = data_tuple[1]
        return menu_info, menu_img
    except Exception as e:
        print(f"Error occurred: {e}")
        return np.nan, np.nan

# 메뉴 정보 데이터 정리
def modify_menu_info(menu_info):
    try:
        # 각 메뉴 정보 문자열을 정규식을 사용하여 수정
        modified_info = [re.sub(r'//', ' ', item) for item in menu_info]
        return modified_info
    except Exception as e:
        print(f"Error occurred: {e}")
        return menu_info

# 키워드 리뷰 칼럼 생성
def add_review_keywords_data(df):
    temp_df = pd.DataFrame()
    for idx, row in df.iterrows():
        try:
            # Convert the string representation of list of dictionaries to actual list of dictionaries
            data_list = eval(row['review_keywords_data'])
            for item in data_list:
                keyword = item['keyword']
                count = int(item['count'].split('인원')[-1])
                temp_df.loc[idx, keyword] = count  # Use .loc to assign values to temp_df
        except Exception as e:
            print(f"Error processing row {idx}: {e}")
    
    # Fill NaN values with 0 and concatenate with original dataframe
    temp_df = temp_df.fillna(0).astype(int)  # Ensure integers
    return pd.concat([df, temp_df], axis=1)

# seats 칼럼 데이터 정리 (1인석 칼럼 원핫인코딩, 최대 인원수 칼럼 생성)
def process_and_extract(df):
    # 함수 정의
    def process_seats(seat):
        words = [word.strip() for word in seat.split(',')]
        processed = []
        contains_1인석 = False

        for word in words:
            if '1인석' in word:
                processed.append('1인석')
                contains_1인석 = True
            elif '최대' in word:
                # '최대' 이전 내용은 모두 삭제
                idx = word.find('최대')
                if idx != -1:
                    processed.append(word[idx:].strip())
                else:
                    processed.append('')
            elif '이하' in word:
                # '단체석' 혹은 '룸'이 포함된 경우 제거
                if '단체석' in word:
                    word = word.replace('단체석', '').strip()
                if '룸' in word:
                    word = word.replace('룸', '').strip()
                processed.append(word)
            else:
                processed.append('')

        # 빈 문자열을 'NaN'으로 변경
        processed_seats = ', '.join([p for p in processed if p]) if processed else 'NaN'

        return processed_seats, contains_1인석

    # 'seats' 칼럼 처리
    df['processed_seats'], df['contains_1인석'] = zip(*df['seats'].apply(process_seats))

    # '1인석' 칼럼 추가
    df['1인석'] = np.where(df['contains_1인석'], 1, 0)

    # 숫자 추출 함수 정의
    def extract_numbers(text):
        numbers = re.findall(r'\d+', text)
        if numbers:
            max_number = max(map(int, numbers))
            if max_number == 1:
                return np.nan
            else:
                return max_number
        else:
            return np.nan

    # processed_seats 칼럼에서 가장 큰 숫자 추출
    df['max_seats'] = df['processed_seats'].apply(lambda x: extract_numbers(x)).astype(float)    
    
    return df['max_seats']

# 새로운 분류 기준 칼럼 생성
final_df['new_category'] = final_df['category'].apply(map_category)

# 데이터프레임의 'distance' 칼럼에 정규식 적용
final_df['distance'] = final_df['distance'].apply(clean_distance)

# 데이터프레임에 적용하여 새로운 칼럼 생성
final_df['website_links'] = final_df['website_links'].apply(extract_href_links)

# 서비스 목록 통합 칼럼 생성
final_df['unique_services'] = final_df.apply(merge_and_remove_duplicates, axis=1)
# home_service와 services 칼럼 삭제
final_df.drop(columns=['home_service', 'services'], inplace=True)

# 부가서비스 원핫인코딩
final_df = create_service_columns(final_df, service_items)
final_df.drop(columns=['unique_services'], inplace=True)

# 주차서비스 원핫인코딩
final_df = create_parking_columns(final_df, parking_items)
final_df.drop(columns=['parking'], inplace=True)

# 메뉴, 메뉴판 이미지 칼럼 생성
final_df[['menu_info', 'menu_img']] = final_df['menu_contents'].apply(extract_menu_info_and_img).apply(pd.Series)
final_df.drop(columns=['menu_contents'], inplace=True)

# menu_info 열의 각 리스트 요소에 대해 함수 apply
final_df['menu_info'] = final_df['menu_info'].apply(lambda x: modify_menu_info(x) if isinstance(x, list) else x)

# 키워드 리뷰 칼럼 생성
final_df = add_review_keywords_data(final_df)

# 'seats' 칼럼 처리
process_and_extract(final_df)
final_df = final_df.drop(columns=['processed_seats', 'contains_1인석'])
```

# 리뷰 텍스트 요약
- OpenAI의 API를 이용하여 리뷰 텍스트 요약

```python
import pandas as pd
import openai
import pandas as pd

final_df = pd.read_csv('./0624_final_df.csv')

# OpenAI API 키 설정
openai.api_key = 'API_KEY_입력'

review = final_df.review_text_list

# 리뷰 요약 함수
def summarize_review(review):
    # 프롬프트 생성
    prompt = (
        "다음의 리뷰를 요약해줘:\n\n"
        f"리뷰: {review}\n\n"
        "1 긍정적인 부분 요약 200토큰 이내로 요약\n"
        "2. 부정적인 부분 요약 200토큰 이내로 요약:\n"
        "3. 주력 메뉴와 그 맛에 대한 사람들의 공통적인 평가 요약 (200토큰 이내)\n"
    )

    # OpenAI API 호출
    response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a helpful assistant who summarizes the restaurant's reviews."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=1000,
        n=1,
        stop=None,
        temperature=0.1
    )

   # 응답 텍스트 추출
    summary = response.choices[0].message['content'].strip()
    return summary

final_df['review_summarized'] = ""
total_reviews = len(final_df)


# 각 리뷰에 대해 요약 수행
for i in range(len(final_df)):
    review = final_df.at[i, 'review_text_list']
    summary = summarize_review(review)
    final_df.at[i, 'review_summarized'] = summary
    
    print(f"Processed {i+1}/{total_reviews} reviews")
    final_df.to_csv('0624_final_df_with_summaries.csv', index=False, encoding='utf-8-sig')
```
