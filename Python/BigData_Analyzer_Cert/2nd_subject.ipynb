## 0) 데이터 불러오기

```python
import requests

# path = r'dataset\customer_test.csv' # test
path = r'dataset\customer_train.csv' # train

# URL
# url = "https://dataq.goorm.io/input-file/content?quizIndex=q_Dlldj_1730448300236&fileName=data%2Fcustomer_test.csv"
url = "https://dataq.goorm.io/input-file/content?quizIndex=q_Dlldj_1730448300236&fileName=data%2Fcustomer_train.csv"
# 헤더나 다른 필요 데이터가 있다면 아래처럼 추가
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
}

# 요청 보내기
response = requests.get(url, headers=headers)

# 상태 코드 확인 (200: 성공)
if response.status_code == 200:
    # CSV 데이터 출력 (파일로 저장 가능)
    with open(path, "wb") as file:
        file.write(response.content)
    print("CSV 파일이 성공적으로 저장되었습니다.")
else:
    print(f"요청 실패: {response.status_code}")
```
