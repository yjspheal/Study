```python
import pandas as pd

train = pd.read_csv(r'dataset\customer_train.csv')
test = pd.read_csv(r'dataset\customer_test.csv')

train
```

## <a id='toc3_2_'></a>[데이터 유형 파악](#toc0_)

```python
train.info()
test.info()
```

## <a id='toc3_3_'></a>[전처리](#toc0_)

### <a id='toc3_3_1_'></a>[(0) 편법으로 빨리하기](#toc0_)

```python
# (1)
```

### <a id='toc3_3_2_'></a>[(1) X, y, train/test 분리](#toc0_)

```python
X_train = train.drop(columns=['회원ID', '총구매액'])
y_train = train['총구매액']
X_test = test.drop(columns = ['회원ID'])
```

```python
print(X_train.shape, X_test.shape, y_train.shape)
```

### <a id='toc3_3_3_'></a>[(2) 결측치 처리](#toc0_)

```python
X_train['환불금액'] = X_train['환불금액'].fillna(0)
X_test['환불금액'] = X_test['환불금액'].fillna(0)
```

```python
print(X_train.isna().sum())
print(X_test.isna().sum())
```

### <a id='toc3_3_4_'></a>[(3) 수치형 변수 스케일링](#toc0_)

```python
X_train.head(3)
```

```python
X_train.columns
```

```python
num_cols = ['최대구매액', '환불금액', '방문일수', '방문당구매건수', '주말방문비율', '구매주기']
cat_cols = ['주구매상품', '주구매지점']
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

scaler = MinMaxScaler()

X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])         # ※fit_transform 아님 주의

X_train.head(3)
```

### <a id='toc3_3_5_'></a>[(4) 범주형 변수 인코딩](#toc0_)

```python
cat_cols
```

```python
# train에 없는 category가 test에 있을 시 확인

for cc in cat_cols:
    print(cc)
    print('train - test **************', set(X_train[cc]) - set(X_test[cc]))
    print('test - train **************',set(X_test[cc]) - set(X_train[cc]))
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

for cc in cat_cols:
    X_train[cc] = encoder.fit_transform(X_train[cc])
    X_test[cc] = encoder .transform(X_test[cc])


X_train.head()
```

## <a id='toc3_4_'></a>[데이터 분리(train,val)](#toc0_)

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size = 0.2, random_state = 42)
# 분류의 경우 train_test_split(X_train, y_train, test_size = 0.2, stratify=y)
print(X_train.shape, X_val.shape, y_train.shape, y_val.shape)
```

## <a id='toc3_5_'></a>[모델 학습 및 검증](#toc0_)

```python
from sklearn.ensemble import RandomForestRegressor
# 분류의 경우 RandomForestClassifier

model = RandomForestRegressor()

model.fit(X_train,y_train)
y_val_pred = model.predict(X_val)

# # 분류의 경우 예측 결과 숫자를 다시 문자열로 변환
# predictions_str = label_encoder.inverse_transform(predictions)
```

## <a id='toc3_6_'></a>[평가](#toc0_)

```python
from sklearn.metrics import root_mean_squared_error, r2_score

rmse = root_mean_squared_error(y_val, y_val_pred)
r2 = r2_score(y_val, y_val_pred)

print(rmse, r2)
```

## <a id='toc3_7_'></a>[결과 제출](#toc0_)

```python
y_pred = model.predict(X_test)

result = pd.DataFrame(y_pred, columns = ['pred'])

result.to_csv('result.csv', index = False)
```
